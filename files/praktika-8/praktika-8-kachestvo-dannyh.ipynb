```python
import great_expectations as gx
import pandas as pd
from datetime import datetime

df = pd.read_csv("calls_sample_data_homework.csv", encoding='utf-8-sig')

# Преобразование столбцов с датами
df['call_start'] = pd.to_datetime(df['call_start'])
df['call_end'] = pd.to_datetime(df['call_end'])

# Расчет длительности звонка в минутах
df['call_duration_minutes'] = (df['call_end'] - df['call_start']).dt.total_seconds() / 60
```

```python
context = gx.get_context()

# Создаем валидатор напрямую из DataFrame
validator = gx.validator(
    execution_engine="pandas",
    data=df,
    expectation_suite_name="calls_suite"
)
```
